```python
# Qiskit Fall Fest 2026
# Challenge 4: Quantum Correlation Switch

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

# Create quantum simulator
simulator = AerSimulator()

# ==========================================
# PART A: CORRELATED OUTCOMES
# Expected: 00 and 11
# ==========================================

qc_A = QuantumCircuit(2, 2)

# Create Bell state
qc_A.h(0)
qc_A.cx(0, 1)

# Measure both qubits
qc_A.measure([0, 1], [0, 1])

print("PART A: CORRELATED OUTCOMES")
display(qc_A.draw(output="mpl"))

# Execute 100 shots
job_A = simulator.run(qc_A, shots=100)
result_A = job_A.result()
counts_A = result_A.get_counts()

print("Part A Measurement Results:")
print(counts_A)

# ==========================================
# PART B: OPPOSITE OUTCOMES
# Expected: 01 and 10
# ==========================================

qc_B = QuantumCircuit(2, 2)

# Create Bell state
qc_B.h(0)
qc_B.cx(0, 1)

# Apply X gate to qubit 0
qc_B.x(0)

# Measure both qubits
qc_B.measure([0, 1], [0, 1])

print("\nPART B: OPPOSITE OUTCOMES")
display(qc_B.draw(output="mpl"))

# Execute 100 shots
job_B = simulator.run(qc_B, shots=100)
result_B = job_B.result()
counts_B = result_B.get_counts()

print("Part B Measurement Results:")
print(counts_B)

# ==========================================
# PART C: COMPARE HISTOGRAMS
# ==========================================

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

plot_histogram(
    counts_A,
    ax=axes[0],
    title="Part A: Correlated Outcomes"
)

plot_histogram(
    counts_B,
    ax=axes[1],
    title="Part B: Opposite Outcomes"
)

plt.tight_layout()
plt.show()

# ==========================================
# PREDICTIONS
# ==========================================

print("\nPREDICTIONS")
print("Part A: 00 and 11 should dominate.")
print("Part B: 01 and 10 should dominate.")

# ==========================================
# COMPARISON AND EXPLANATION
# ==========================================

print("\nCOMPARISON")
print("Part A Results:", counts_A)
print("Part B Results:", counts_B)

print("\nEXPLANATION")
print(
    "Part A: The Bell state produces correlated measurement "
    "outcomes, predominantly 00 and 11."
)

print(
    "Part B: Applying an X gate to qubit 0 flips that qubit, "
    "changing the outcomes to predominantly 01 and 10."
)

print(
    "Both circuits use exactly 2 qubits and 100 shots. "
    "Each circuit uses no more than 4 gates before measurement."
)
```
